```python
import pandas as pd
list_of_tables=["C:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_customers_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_geolocation_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_order_items_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_order_payments_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_order_reviews_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_orders_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_products_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_sellers_dataset.csv",
"c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/product_category_name_translation.csv"]
name_table=["olist_customers","olist_geolocation","olist_order_items","olist_order_payments","olist_order_reviews"
,"olist_orders","olist_products","olist_sellers","product_category_name_translation"]
summary=[]
for i in range(len(list_of_tables)):
    df=pd.read_csv(list_of_tables[i])
    table_info={
        "table_name":name_table[i],
        "rows":df.shape[0],
        "columns":df.shape[1],
        "total_null":df.isnull().sum().sum(),
        "duplicates":df.duplicated().sum()
    }
    summary.append(table_info)
table_sumary=pd.DataFrame(summary)
Column_Level_Summary=[]
for i in range(len(list_of_tables)):
    df=pd.read_csv(list_of_tables[i])
    list_columns=list(df.columns)
    for j in range(df.shape[1]):
        column_info={
            "Table":name_table[i],
            "Columns":list_columns[j],
            "Data_Type":df[list_columns[j]].dtype,
            "Nulls":df[list_columns[j]].isnull().sum(),
            "Null_Percentage":((df[list_columns[j]].isnull().sum())/(df.shape[0]))*100,
            "unique":df[list_columns[j]].nunique()
        }
        Column_Level_Summary.append(column_info)
table_columns=pd.DataFrame(Column_Level_Summary)    
print(f"{table_sumary}\n---------------------\n")    
print(table_columns)




```

## 3.1 Duplicate Analysis
### Geolocation Duplicate Investigation

```python
df=pd.read_csv("c:/Users/Adora IT/Desktop/olist/olist-data-analysis/data/olist_geolocation_dataset.csv")
print(df.duplicated().sum())
print(df['geolocation_zip_code_prefix'].nunique())
info=df.groupby("geolocation_zip_code_prefix").agg(
    records=('geolocation_zip_code_prefix','count'),
    unique_lat=('geolocation_lat','nunique'),
    unique_lng=('geolocation_lng', 'nunique') 

).reset_index()
print(info)
```

```python
count_without_duplicates = df.drop_duplicates(inplace=False).shape[0]
print(count_without_duplicates)
```

Geolocation Duplicate Analysis:
The geolocation table contains 1,000,163 records and 19,015 unique ZIP code prefixes. A ZIP code prefix can correspond to multiple geographic coordinates, so repeated ZIP prefixes are not considered duplicates by themselves. However, 261,831 complete duplicate rows were identified. Removing these complete duplicates would leave 738,332 unique records. These duplicates will be removed during the data-cleaning phase
